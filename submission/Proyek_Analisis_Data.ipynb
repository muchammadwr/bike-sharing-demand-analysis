## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Question 1:** During the 2012 period, which days and hours recorded the highest average bike rentals?

- **Question 2:** How did monthly rental demand change throughout the 2012 period?

- **Question 3:** How did average bike rentals differ across weather conditions during 2012?

- **Question 5:** How did rental patterns differ between casual and registered users during 2012?

### Cleaning Data

#### Fixing daily_df problem

```python
daily_df.head()
```

```python
# Change the column names
daily_columns = {
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "temp": "temperature",
    "atemp": "feels_like_temperature",
    "hum": "humidity",
    "cnt": "count",
}

daily_df.rename(columns=daily_columns, inplace=True)
```

```python
# Fix the datetime data
daily_df["date"] = pd.to_datetime(daily_df["date"])
```

```python
# Categorical season column
season = {1: "Springer", 2: "Summer", 3: "Fall", 4: "Winter"}

daily_df["season"] = daily_df["season"].replace(season)
```

```python
# Year column
year = {0: 2011, 1: 2012}

daily_df["year"] = daily_df["year"].replace(year)
```

```python
# Month column
month = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December",
}

daily_df["month"] = daily_df["month"].replace(month)
```

```python
# Change holiday column to categorical values
holiday = {0: "No", 1: "Yes"}

daily_df["holiday"] = daily_df["holiday"].replace(holiday)
```

```python
# Change coloumns weekdays from numerical to categorical
weekdays = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday",
}

daily_df["weekday"] = daily_df["weekday"].replace(weekdays)
```

```python
# Change coloumns workingday from numerical to categorical
workingday = {0: "No", 1: "Yes"}

daily_df["workingday"] = daily_df["workingday"].replace(workingday)
```

```python
# Change the weathersit from numerical to categorical
weathersit_labels = {
    1: "Clear",
    2: "Cloudy or Misty",
    3: "Light Rain/Snow",
    4: "Heavy Rain/Snow",
}

daily_df["weathersit"] = daily_df["weathersit"].map(weathersit_labels)
```

```python
# Denormalized temperature column with multiple by 41
daily_df["temperature"] = daily_df["temperature"] * 41
```

```python
# Denormalized feels_like_temperature column with multiple by 50
daily_df["feels_like_temperature"] = daily_df["feels_like_temperature"] * 50
```

```python
# Denormalized windspeed column with multiple by 67
daily_df["windspeed"] = daily_df["windspeed"] * 67
```

## Visualization & Explanatory Analysis

### **Question 1:** During the 2012 period, which days and hours recorded the highest average bike rentals?


```python
# Total & Average Rentals
daily_2012 = daily_df[daily_df['year'] == 2012]

total_rentals = daily_2012["count"].sum()
average_rentals = daily_2012["count"].mean()

print(f"Total Rentals: {total_rentals:,.0f}")
print(f"Average Rentals: {average_rentals:,.0f}")
```

#### Daily

```python
# Average and Total rentals by days in 2012
day_order = [
    "Sunday", "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday"
]

daily_2012 = daily_df[daily_df['year'] == 2012]
daily_stats_2012 = daily_2012.groupby('weekday')['count'].agg(['mean', 'sum']).round(2).reset_index()
daily_stats_2012.columns = ['days', 'average_rentals', 'total_rentals']
daily_stats_2012.sort_values(by='average_rentals', ascending=False)
```

```python
# Visualization of Average Rentals by Hours in 2012 - Line Chart
fig, ax = plt.subplots(figsize=(10, 6))
sns.set_style("white")

# Line chart
sns.lineplot(data=daily_stats_2012, x='days', y='average_rentals', marker='o', color='#72BCD4', linewidth=2.5, ax=ax)

ax.set_xlabel('Days', fontsize=12)
ax.set_ylabel('Average Rentals', fontsize=12)

plt.title('Average Rentals by Days (2012)', fontsize=16, pad=20)

ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Visualization of Total Rentals by Hours in 2012 - Line Chart
fig, ax = plt.subplots(figsize=(10, 6))
sns.set_style("white")

# Line chart
sns.lineplot(data=daily_stats_2012, x='days', y='total_rentals', marker='o', color='#72BCD4', linewidth=2.5, ax=ax)

ax.set_xlabel('Days', fontsize=12)
ax.set_ylabel('Total Rentals', fontsize=12)

plt.title('Total Rentals by Days (2012)', fontsize=16, pad=20)

ax.legend()
plt.tight_layout()
plt.show()
```

**Insight — Rental Demand by Day of Week**

* **Thursday has the highest rental demand**, with **5,978 average rentals** and **310,843 total rentals**.

### **Question 5:** How did rental patterns differ between casual and registered users during 2012?


```python
# Filter 2012 Users

daily_2012 = daily_df[daily_df['year'] == 2012]
user_weekday = (
    daily_2012.groupby("weekday")[["casual", "registered"]]
    .mean()
    .reindex(
        ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
    )
    .round(2)
    .reset_index()
)

user_weekday
```

```python
user_weekday = user_weekday.melt(
    id_vars="weekday",
    value_vars=["casual", "registered"],
    var_name="user_type",
    value_name="average_rentals",
)
user_weekday
```

```python
user_weekday['user_type'] = user_weekday['user_type'].map({'casual': 'Casual', 'registered': 'Registered'})
```

```python
# Visualization rental patterns differ between casual and registered users during 2012?

fig, ax = plt.subplots(figsize=(10, 6))

sns.barplot(
    data=user_weekday,
    x="weekday",
    y="average_rentals",
    hue="user_type",
    palette={"Registered": "#72BCD4", "Casual": "#D3D3D3"},
    ax=ax,
)

ax.set_title("Average Bike Rentals by User Type and Weekday")
ax.set_xlabel("Days")
ax.set_ylabel("Average Rentals")
ax.tick_params(axis="x", rotation=30)

ax.legend(title="User Type", loc="upper right")
ax.grid(False)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.lineplot(
    data=user_weekday,
    x="weekday",
    y="average_rentals",
    hue="user_type",
    marker="o",
    linewidth=2,
    palette={"Registered": "#72BCD4", "Casual": "#D3D3D3"},
    ax=ax,
)

ax.set_title("Rental Patterns by User Type and Weekday")
ax.set_xlabel("Weekday")
ax.set_ylabel("Average Rentals")

ax.tick_params(axis="x", rotation=45)

ax.legend(title="User Type", loc="upper right")

plt.tight_layout()
plt.show()
```

### **Question 6:** Which user segment contributed the largest share of total rentals over the two-year period?

```python
user_share = daily_df[["casual", "registered"]].sum().to_frame("total_rentals")

user_share["rental_share"] = (
    user_share["total_rentals"] / user_share["total_rentals"].sum() * 100
).round(2)

user_share
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

colors = ["#D3D3D3", "#72BCD4"]

# Create pie chart without labels or autopct inside the wedges
ax.pie(user_share["rental_share"], startangle=90, colors=colors)

# Adding requested visual elements
ax.set_title("Rental Contribution by User Type", fontsize=14, pad=20)
ax.set_xlabel("Share Distribution")
ax.set_ylabel("User Categories")

# Add legend using the index (Casual, Registered)
ax.legend(labels=[i.title() for i in user_share.index], loc="upper right", title="User Type")

plt.tight_layout()
plt.show()
```
