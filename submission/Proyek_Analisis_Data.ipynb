## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Question 1:** During the 2012 period, which hours, days, and seasons recorded the highest average bike rentals?

- **Question 2:** How did monthly rental demand change throughout the 2012 period?

- **Question 3:** How did average bike rentals differ across weather conditions during 2012?

- **Question 4:** Which weather-related factor (temperature, humidity, or wind speed) had the strongest relationship with rental demand?

- **Question 5:** How did rental patterns differ between casual and registered users during 2012?

- **Question 6:** Which user segment contributed the largest share of total rentals over the two-year period?

The dataset contains various columns, including:

- **instant**: Record index.

- **hum**: Normalized humidity. The values are divided to 100 (max).
- **windspeed**: Normalized wind speed. The values are divided to 67 (max).

- **registered**: Count of registered users.
- **cnt**: Count of total rental bikes including both casual and registered.

### Cleaning Data

#### Fixing daily_df problem

```python
daily_df.head()
```

```python
# Change the column names
daily_columns = {
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "temp": "temperature",
    "atemp": "feels_like_temperature",
    "hum": "humidity",
    "cnt": "count",
}

daily_df.rename(columns=daily_columns, inplace=True)
```

```python
# Fix the datetime data
daily_df["date"] = pd.to_datetime(daily_df["date"])
```

```python
# Categorical season column
season = {1: "Springer", 2: "Summer", 3: "Fall", 4: "Winter"}

daily_df["season"] = daily_df["season"].replace(season)
```

```python
# Year column
year = {0: 2011, 1: 2012}

daily_df["year"] = daily_df["year"].replace(year)
```

```python
# Month column
month = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December",
}

daily_df["month"] = daily_df["month"].replace(month)
```

```python
# Change holiday column to categorical values
holiday = {0: "No", 1: "Yes"}

daily_df["holiday"] = daily_df["holiday"].replace(holiday)
```

```python
# Change coloumns weekdays from numerical to categorical
weekdays = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday",
}

daily_df["weekday"] = daily_df["weekday"].replace(weekdays)
```

```python
# Change coloumns workingday from numerical to categorical
workingday = {0: "No", 1: "Yes"}

daily_df["workingday"] = daily_df["workingday"].replace(workingday)
```

```python
# Change the weathersit from numerical to categorical
weathersit_labels = {
    1: "Clear",
    2: "Cloudy or Misty",
    3: "Light Rain/Snow",
    4: "Heavy Rain/Snow",
}

daily_df["weathersit"] = daily_df["weathersit"].map(weathersit_labels)
```

```python
# Denormalized temperature column with multiple by 41
daily_df["temperature"] = daily_df["temperature"] * 41
```

```python
# Denormalized feels_like_temperature column with multiple by 50
daily_df["feels_like_temperature"] = daily_df["feels_like_temperature"] * 50
```

```python
# Denormalized windspeed column with multiple by 67
daily_df["windspeed"] = daily_df["windspeed"] * 67
```

Based on the descriptive statistics:

- The average daily bike rentals were **4,504**, consisting of approximately **848 casual users** and **3,656 registered users**.

```python
# numerical column values
num = daily_df.select_dtypes(
    exclude=["object", "category", "datetime"]
).columns.tolist()
daily_df[num].head()
```

```python
# datetype column values
num = daily_df.select_dtypes(include=["datetime"]).columns.tolist()
daily_df[num].head()
```

```python
# categorical column values
num = daily_df.select_dtypes(
    include=["object", "category"]
).columns.tolist()
daily_df[num].head()
```

```python
# Rented by season
daily_df.groupby("season")["count"].agg({"sum", "mean", "count"}).round(2)
```

Based on the seasonal summary:

- **Fall** recorded the highest average daily rentals (**5,644**), indicating the strongest demand among all seasons.
- **Spring** had the lowest average daily rentals (**2,604**), suggesting lower bike usage during this season.
- **Summer** and **Winter** showed relatively similar rental demand, averaging **4,992** and **4,728** daily rentals, respectively.
- Rental variability (standard deviation) was highest in **Winter** and **Summer**, indicating more fluctuation in daily demand, while **Fall** exhibited the most consistent rental patterns.


```python
# Rented by month
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

daily_df["month"] = pd.Categorical(daily_df["month"], categories=month_order, ordered=True)
daily_df.groupby("month")["count"].agg({"sum", "mean", "count"}).round(2)
```

```python
#  Rented by Holiday
daily_df.groupby("holiday")["count"].agg({"sum", "mean", "count"}).round(2)
```

```python
# Rented by Weekday
day_order = [
    "Sunday", "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday"
]

daily_df["weekday"] = pd.Categorical(daily_df["weekday"], categories=day_order, ordered=True)
daily_df.groupby("weekday")["count"].agg({"sum", "mean", "count"}).round(2)
```

Based on the weekday summary:

- **Friday** recorded the highest average daily rentals (**4,690**), followed closely by **Thursday** (**4,667**).
- **Sunday** had the lowest average daily rentals (**4,229**), indicating lower bike usage on weekends.
- Total rentals were relatively consistent across weekdays, ranging from **444,027** (Sunday) to **487,790** (Friday).
- Overall, bike rental demand remained fairly stable throughout the week, with slightly higher activity on weekdays than weekends.


```python
# rented by month
daily_df.groupby("month")["count"].agg({"sum", "mean", "count"}).round(2)
```

Based on the monthly summary:

- **June** recorded the highest average daily rentals (**5,772**), followed closely by **September** (**5,766**) and **August** (**5,664**).
- **January** had the lowest average daily rentals (**2,176**), with **February** also showing relatively low demand (**2,655**).
- Bike rental demand generally increased from **spring to summer**, peaked during the warmer months, and gradually declined toward the end of the year.
- These results suggest that **seasonal and weather conditions** play an important role in influencing bike rental demand.


```python
# Sales performance by years
daily_df.groupby("year")["count"].agg({"sum", "mean", "count"}).round(2)
```

- **2012** recorded significantly higher bike rental demand than **2011**, with an average of **5,600** daily rentals compared to **3,406** in 2011.
- Total rentals increased from **1.24 million** in **2011** to **2.05 million** in **2012**, representing substantial growth in bike-sharing usage.
- These results indicate a strong upward trend in demand, suggesting increased adoption and utilization of the bike-sharing service over the two-year period.


```python
# registered customers compare to casual customers
daily_df.groupby("weekday").agg(
    {"casual": ["mean", "sum"], "registered": ["mean", "sum"]}
).round(2)
```

Based on the weekday comparison:

- **Registered users** consistently contributed more rentals than **casual users** on every day of the week, indicating they are the primary customer segment.
- **Casual users** were most active on **Saturday** (average **1,465** rentals) and **Sunday** (average **1,338** rentals), suggesting higher recreational usage during weekends.
- **Registered users** recorded their highest average rentals on **Thursday** (**4,076**) and **Friday** (**3,938**), reflecting strong commuting demand on weekdays.
- These patterns indicate that **registered users primarily use the service for weekday commuting**, while **casual users are more likely to rent bikes for leisure on weekends**.
