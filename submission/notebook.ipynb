### Cleaning Data


#### Fixing hourly_data problem


```python
hourly_df.head()
```

```python
# Change the column names
hourly_column = {
    "dteday": "date",
    "yr": "year",
    "mnth": "month",
    "hr": "hour",
    "temp": "temperature",
    "atemp": "feels_like_temperature",
    "hum": "humidity",
    "cnt": "count",
}

hourly_df.rename(columns=hourly_column, inplace=True)
hourly_df.head()
```

```python
# Fix the datetime data
hourly_df["date"] = pd.to_datetime(hourly_df["date"])
```

```python
# Categorical season column
season = {1: "Springer", 2: "Summer", 3: "Fall", 4: "Winter"}

hourly_df["season"] = hourly_df["season"].replace(season)
```

```python
# Year column
year = {0: 2011, 1: 2012}

hourly_df["year"] = hourly_df["year"].replace(year)
```

```python
# Month column
month = {
    1: "January",
    2: "February",
    3: "March",
    4: "April",
    5: "May",
    6: "June",
    7: "July",
    8: "August",
    9: "September",
    10: "October",
    11: "November",
    12: "December",
}

hourly_df["month"] = hourly_df["month"].replace(month)
```

```python
# Hour Column
hourly_df["hour"] = hourly_df["hour"].apply(lambda x: time(x, 0))
```

```python
hourly_df.head()
```

```python
# Change holiday column to categorical values
holiday = {0: "No", 1: "Yes"}

hourly_df["holiday"] = hourly_df["holiday"].replace(holiday)
```

```python
# Change coloumns weekdays from numerical to categorical
weekdays = {
    0: "Sunday",
    1: "Monday",
    2: "Tuesday",
    3: "Wednesday",
    4: "Thursday",
    5: "Friday",
    6: "Saturday",
}

hourly_df["weekday"] = hourly_df["weekday"].replace(weekdays)
```

```python
# Change coloumns workingday from numerical to categorical
workingday = {0: "No", 1: "Yes"}

hourly_df["workingday"] = hourly_df["workingday"].replace(workingday)
```

```python
# Change the weathersit from numerical to categorical
weathersit_labels = {
    1: "Clear",
    2: "Cloudy or Misty",
    3: "Light Rain/Snow",
    4: "Heavy Rain/Snow",
}

hourly_df["weathersit"] = hourly_df["weathersit"].map(weathersit_labels)
```

```python
# Denormalized temperature column with multiple by 41
hourly_df["temperature"] = hourly_df["temperature"] * 41
```

```python
# Denormalized feels_like_temperature column with multiple by 50
hourly_df["feels_like_temperature"] = hourly_df["feels_like_temperature"] * 50
```

```python
# Denormalized humidity column with multiple by 100
hourly_df["humidity"] = hourly_df["humidity"] * 100
```

## Visualization & Explanatory Analysis


### **Question 1:** During the 2012 period, which hours, days, and seasons recorded the highest average bike rentals?

```python
# Total & Average Rentals
daily_2012 = daily_df[daily_df['year'] == 2012]

total_rentals = daily_2012["count"].sum()
average_rentals = daily_2012["count"].mean()

print(f"Total Rentals: {total_rentals:,.0f}")
print(f"Average Rentals: {average_rentals:,.0f}")
```

#### Hourly

```python
# Average and Total rentals by hours in 2012

hourly_2012 = hourly_df[hourly_df['year'] == 2012]
hourly_stats_2012 = hourly_2012.groupby('hour')['count'].agg(['mean', 'sum']).round(2).reset_index()
hourly_stats_2012.columns = ['hour', 'average_rentals', 'total_rentals']
hourly_stats_2012.sort_values('average_rentals', ascending=False)
```

```python
# Visualization of Average Rentals by Hours in 2012

hourly_stats_2012['hour_int'] = hourly_stats_2012['hour'].apply(lambda x: x.hour)

fig, ax = plt.subplots(figsize=(10, 6))
sns.set_style("white")

# Line chart
sns.lineplot(data=hourly_stats_2012, x='hour_int', y='average_rentals', marker='o', color='#72BCD4', linewidth=2.5, ax=ax)

ax.set_xlabel('Hours', fontsize=12)
ax.set_ylabel('Average Rentals', fontsize=12)

plt.title('Average Rentals by Hours (2012)', fontsize=16, pad=20)
ax.set_xticks(range(0, 24))
ax.set_xticklabels([f"{h:02d}:00" for h in range(24)], rotation=30)
ax.grid(False)

ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Visualization of Total Rentals by Hours in 2012

hourly_stats_2012['hour_int'] = hourly_stats_2012['hour'].apply(lambda x: x.hour)

fig, ax = plt.subplots(figsize=(10, 6))
sns.set_style("white")

# Line chart
sns.lineplot(data=hourly_stats_2012, x='hour_int', y='total_rentals', marker='o', color='#72BCD4', linewidth=2.5, ax=ax)

ax.set_xlabel('Hours', fontsize=12)
ax.set_ylabel('Total Rentals', fontsize=12)

plt.title('Total Rentals by Hours (2012)', fontsize=16, pad=20)
ax.set_xticks(range(0, 24))
ax.set_xticklabels([f"{h:02d}:00" for h in range(24)], rotation=45)

ax.legend()
plt.tight_layout()
plt.show()
```

**Insight — Rental Demand by Hour in 2012**

* **17:00 has the highest rental demand**, with an average of **573 rentals** and **209,228 total rentals**.
* **18:00 is the second-highest**, with an average of **528 rentals**.
* Rental demand is **lowest at 04:00**, with only **7 average rentals** and **2,611 total rentals**.
* Demand increases significantly from **06:00**, peaks around **17:00–18:00**, and then gradually decreases at night.
* Overall, **afternoon and early evening are the busiest periods**, while **early morning has the lowest rental activity**.

## Further Analysis


### Manual Grouping

Develop a manual demand grouping based on predefined business rules to categorize hourly rental demand and support better bike availability planning and operational decisions. A clear classification of hourly rental periods into Low, Medium, and High Demand, supported by visualizations and summary statistics.

#### Hourly

```python
# Copy hourly dataset
hourly_grouping_df = hourly_df.copy()
hourly_grouping_df.head()
```

```python
hourly_demand_group_df = hourly_grouping_df.groupby("hour")["count"].agg(["sum", "mean"]).reset_index().round(2)
hourly_demand_group_df.columns = ["hour", "total", "average"]
hourly_demand_group_df
```

```python
# Define thresholds based on hourly quartiles mean
q1_hourly_avg = hourly_demand_group_df["average"].quantile(0.25).round(2)
q3_hourly_avg = hourly_demand_group_df["average"].quantile(0.75).round(2)

print(f"Low Demand threshold (Hourly): {q1_hourly_avg}")
print(f"High Demand threshold (Hourly): {q3_hourly_avg}")
```

```python
# Create hourly classification function
def classify_hourly_demand_group(average):
    if average <= q1_hourly_avg:
        return "Low Demand"
    elif average <= q3_hourly_avg:
        return "Medium Demand"
    else:
        return "High Demand"
```

```python
# Apply classification
hourly_demand_group_df["demand_group"] = hourly_demand_group_df["average"].apply(classify_hourly_demand_group)
hourly_demand_group_df[["hour", "total","average",  "demand_group"]]
```

```python
hourly_demand_group_distribution = hourly_demand_group_df.groupby("demand_group")["total"].sum().round(2).reindex(["Low Demand", "Medium Demand", "High Demand"]).reset_index()
hourly_demand_group_distribution
```

```python
# Hourly Total Demands Visualization
fig, ax = plt.subplots(figsize=(10, 6))
colors = ["#D3D3D3", "#55A868", "#D3D3D3"]

sns.barplot(
    data=hourly_demand_group_distribution,
    x="demand_group",
    y="total",
    hue="demand_group",
    palette=colors,
    legend=False,
    ax=ax
)

plt.title(
    "Hourly Bike Rental Demand Classification",
    fontsize=16
)

ax.grid(False)

plt.xlabel("Demand Group")
plt.ylabel("Number of Hours")

plt.tight_layout()
plt.show()
```

```python
# Visualization demand groups by hours

# Color mapping for each demand group
demand_colors = {
    "Low Demand": "#F4A6B8",     # Pink
    "Medium Demand": "#BDBDBD",  # Gray
    "High Demand": "#55A868",    # Green
}


fig, ax = plt.subplots(figsize=(14, 6))

sns.barplot(
    data=hourly_demand_group_df,
    x="hour",
    y="average",
    hue="demand_group",
    hue_order=[
        "Low Demand",
        "Medium Demand",
        "High Demand",
    ],
    palette=demand_colors,
    errorbar=None,
    ax=ax,
)

ax.grid(False)

plt.title(
    "Average Hourly Bike Rental Demand",
    fontsize=16
)

plt.xlabel("Hour")
plt.ylabel("Average Rentals")
plt.xticks(rotation=30)


plt.tight_layout()
plt.show()
```

**Insights:**

* **Low demand** occurs from **00:00–05:00**, reaching its lowest point at **04:00** with **6.35 average rentals**.
* Demand rises sharply during morning commuting hours, reaching **359.01 average rentals at 08:00**.
* Demand peaks at **17:00** with **461.45 average rentals**, followed by **18:00** with **425.51**.
* **High-demand hours** are concentrated around morning and evening commuting periods.
* After **19:00**, demand gradually declines from **311.52** to **87.83 average rentals at 23:00**.
